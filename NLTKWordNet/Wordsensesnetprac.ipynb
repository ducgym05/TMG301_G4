```python
import nltk
from nltk.corpus import wordnet

nltk.download('wordnet')
nltk.download('omw-1.4')  # Hỗ trợ đa ngôn ngữ

print("WordNet đã sẵn sàng!")
```

```python
# Task 1: Khám phá các synset của một từ đa nghĩa
word = "bat"
print(f"\n=== Tất cả synsets của từ '{word}' ===\n")

synsets = wordnet.synsets(word)

for i, synset in enumerate(synsets):
    print(f"Synset {i+1}: {synset.name()}")
    print(f"   Định nghĩa: {synset.definition()}")
    print(f"   Ví dụ: {synset.examples() if synset.examples() else 'Không có ví dụ'}")
    print("-" * 60)
```

```python
# Task 2 & 3:
chosen_synset = wordnet.synsets("bat")[1]
print(f"\n=== Phân tích synset: {chosen_synset.name()} ===")
print(f"Định nghĩa: {chosen_synset.definition()}\n")

# Task 2: Hypernyms Hyponyms
print("1. Hypernyms (siêu loại - khái niệm rộng hơn):")
hypernyms = chosen_synset.hypernyms()
for h in hypernyms:
    print(f"   - {h.name()} : {h.definition()[:100]}...")

print("\n2. Hyponyms (loại con - khái niệm cụ thể hơn):")
hyponyms = chosen_synset.hyponyms()
for h in hyponyms:
    print(f"   - {h.name()} : {h.definition()[:100]}...")

# Task 3: Synonyms Antonyms
print("\n3. Synonyms (từ đồng nghĩa):")
synonyms = set()
for lemma in chosen_synset.lemmas():
    synonyms.add(lemma.name().replace("_", " "))
print(f"   {', '.join(sorted(synonyms))}")

print("\n4. Antonyms (từ trái nghĩa):")
antonyms = set()
for lemma in chosen_synset.lemmas():
    antonyms.update([ant.name().replace("_", " ") for ant in lemma.antonyms()])
if antonyms:
    print(f"   {', '.join(antonyms)}")
else:
    print("   Không có antonym trong synset này.")
```
